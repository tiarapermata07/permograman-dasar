2. Buatlah contoh penggunaan karakter %d, %f, %s, %x, dan %o untuk fungsi output.

```python
# %d untuk integer
umur = 19
print("Umur saya adalah %d tahun" % umur)

# %f untuk float
nilai = 85.5
print("Nilai saya adalah %f" % nilai)

# %s untuk string
nama = "Taya"
print("Nama saya adalah %s" % (nama))

# %x untuk hexadecimal
angka = 255
print("Nilai hexadecimal dari %d adalah %x" % (angka , angka))

# %o untuk octal
angka = 64
print("Nilai octal dari %d adalah %o" % (angka , angka))
```

1. Operator Aritmetika +-

Studi Kasus: Konversi Total Detik

Anda diberi total detik, dan tugas Anda adalah mengubahnya menjadi format Jam:Menit:Detik.

*   Gunakan operator pembagian integer (//) untuk menemukan jumlah jam dan
menit
*   Gunakan operator modulus (%) untuk menemukan sisa detik dan menit.

```python
# konversi waktu
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

2. Buatlah program untuk menghitung nilai akhir mahasiswa berdasarkan nilai tugas, UTS, dan UAS dengan ketentuan sebagai berikut:

* Modifikasi agar variabel nilai_mahasiswa dan standar_kelulusan bisa menerima inputan melalui keyboard.
* Modifikasi program agar nilai mahasiswa terdiri dari 5 variabel dengan input via keyboard.

```python
tugas = 80
uts = 70
uas = 90

nilai_akhir = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)
print("Nilai akhir =", nilai_akhir)
```

```python
tugas = float(input("Masukkan nilai tugas: "))
kuis = float(input("Masukkan nilai kuis: "))
praktikum = float(input("Masukkan nilai praktikum: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))

nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)

standar_kelulusan = float(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

3. Menghitung Diskon Belanja

* Modifikasi inputan via keyboard.
* Modifikasi program agar total diskon yang diberikan menjadi 35% dengan ketentuan belanja di atas Rp1.000.000

```python
belanja = float(input("Masukkan total belanja: "))

if belanja > 1000000:
    diskon = 0.35 * belanja
else:
    diskon = 0

total_harga = belanja - diskon

print("Diskon =", diskon)
print("Total harga setelah diskon =", total_harga)
```
