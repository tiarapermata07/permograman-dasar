1. Menentukan Tahun Kabisat

Studi Kasus
Sebuah tahun disebut tahun kabisat jika memenuhi salah satu dari dua kondisi berikut:

a. Tahun habis dibagi 4 tetapi tidak habis dibagi 100

b. Tahun habis dibagi 400.

Buatlah sebuah ekspresi logika kompleks untuk menentukan apakah suatu tahun merupakan tahun kabisat.

```python
tahun = int(input("Masukkan tahun: "))

if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
    print(tahun, "merupakan tahun kabisat")
else:
    print(tahun, "bukan tahun kabisat")
```

2. Buatlah program sederhana versi anda masing-masing dengan lingkup,fungsi input, fungsi output, dan operator artimatika

```python
def hitung_nilai_akhir(nilai_tugas, nilai_uts, nilai_uas):
    nilai_akhir = (nilai_tugas * 0.3) + (nilai_uts * 0.3) + (nilai_uas * 0.4)
    return nilai_akhir


# Input
nama = input("Masukkan nama: ")
nilai_tugas = float(input("Masukkan nilai tugas: "))
nilai_uts = float(input("Masukkan nilai UTS: "))
nilai_uas = float(input("Masukkan nilai UAS: "))

# Proses
nilai_akhir = hitung_nilai_akhir(nilai_tugas, nilai_uts, nilai_uas)

# Output
print("\nNama:", nama)
print("Nilai akhir:", nilai_akhir)

if nilai_akhir >= 75:
    print("Keterangan: Lulus")
else:
    print("Keterangan: Tidak Lulus")
```
