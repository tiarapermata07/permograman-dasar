Sertakan sumber yang kamu gunakan.

Python memiliki aturan prioritas operator yang menjelaskan urutan pelaksanaan operator yang mana terlebih dahulu dikerjakan:

•	Tanda kurung memiliki prioritas tertinggi, artinya ekspresi di dalam tanda kurung harus dievaluasi terlebih dahulu.

•	Perkalian (*) memiliki prioritas lebih tinggi daripada penjumlahan +, dan oleh karena itu perkalian dievaluasi sebelum penjumlahan.

•	Jika dua operator memiliki prioritas yang sama, ekspresi dievaluasi dari kiri ke kanan. Penjumlahan (+) dan pengurangan (-) memiliki prioritas yang sama, oleh karena itu evaluasi ekspresi dari kiri ke kanan.

penjelasan mengapa ekspresi 10 + 5 * 2 menghasilkan 20
Karena pada tabel di atas dan sesuai penjelasan yang saya dapat bahwa perkalian memiliki prioritas lebih tinggi daripada operator penjumlahan. sehingga ekspresi 10 + 5 * 2 menghasilkan 20.


```python
(10 + 5 * 2)
print("hasilnya adalah", (10 + 5 * 2))
```

Bukan 30, karena perkalian dikerjakan terlebih dahulu. Jadi prosesnya seperti ini:

```python
print("10 + 5 * 2")
print("10 + (5 * 2)")
print("10 + ", (5 * 2) )
print("hasilnya adalah", (10 + 5 * 2))
```

Jika penjumlahan dikerjakan terlebih dahulu untuk menghasilkan 30, kita dapat menggunakan tanda kurung pada bagian penjumlahan:

```python
print("10 + 5 * 2")
print( "(10 + 5) * 2" )
print((10 + 5), "* 2")
print("hasilnya adalah", (10 + 5) * 2)
```

2. Jelaskan perbedaan mendasar antara operator relasional (>, <, ==, dll.) dan operator logika (and, or, not) dalam Python. Berikan satu contoh situasi nyata (boleh di luar konteks pembangkit) di mana kedua jenis operator ini harus dipakai bersamaan dalam satu ekspresi untuk menghasilkan keputusan yang benar.

Operator relasional (>, <, ==, dll.) : Digunakan untuk membandingkan dua nilai dan menghasilkan nilai True atau False

Operator logika (and, or, not) : Digunakan untuk menggabungkan atau membalik kondisi.


Contoh situasi nyata

Menentukan kelulusan siswa berdasarkan nilai dan kkm. Siswa dinyatakan lulus apabila nilai yang diperoleh lebih besar atau sama dengan kkm dan tugas telah dikumpulkan. Situasi ini dapat membuat sebuah program python sederhana dengan menggunakan jenis operator relasional dan operator logika yang dipakai bersamaan dalam satu ekspresi untuk menghasilkan keputusan yang benar.

```python
nilai = 88
kkm = 70
tugas = True

hasil = nilai >= kkm and tugas

print("Nilai =", nilai)
print("KKM =", kkm)
print("Tugas dikumpulkan =", tugas)
print("Lulus =", hasil)
```

•	nilai >= kkm → mengecek apakah nilai sudah mencapai kkm (operator relasional).

•	and → memastikan nilai memenuhi kkm dan tugas sudah dikumpulkan (operator logika).

•	Hasil akhirnya berupa True (benar).

Pengguna masih baru dalam pemrograman python sehingga ia mempertanyakan mengapa dua rumus yang terlihat mirip menghasilkan nilai berbeda Pengguna juga mempertanyakan, “Bukankah prioritas operator (*)lebih tinggi daripada ( / ) dan seharusnya memberikan jawaban yang sama?”


```python
import math

hasil1 = math.log(0.05 / (2 * 0.000001), 10)
hasil2 = math.log(0.05 / 2 * 0.000001, 10)

print(hasil1)
print(hasil2)
```

Kesalahan dapat terjadi karena operator (*) dan ( / ) sebenarnya memiliki prioritas yang sama, sehingga ketika tidak ada tanda kurung, keduanya dikerjakan dari kiri ke kanan. Perbedaan hasil yang didapatkan dapat diperbaiki atau dijelaskan sebagai berikut:

1.	math.log(0.05/(20*000001),10) yang memiliki hasil =4.3979400086720375 menggunakan tanda kurung ( ) dalam perkalian (20*000001) ini. Jadi, sesuai aturan dasar matematika untuk penyederhanaan yaitu BODMAS(Bracket of division multiplication addition subtraction). Kode tersebut pertama-tama mengalikan ( ) bagian dalam tanda kurung, kemudian dilanjutkan dengan pembagian.

2.	math.log(0.05/2*0.000001,10) yang memiliki hasil                                              = -7.602059991327962 tidak ada gunanya diberi ( ) sehingga pertama-tama dilakukan operasi pembagian lalu dikalikan dengan 0.000001.
