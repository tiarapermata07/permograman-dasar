Soal 1 : Belanja Buku

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

1. Total belanja, besar diskon, dan jumlah yang harus dibayar.
2. Uang kembalian.
3. Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
4. Apakah uang Budi cukup? (hasil True/False)

```python
harga_buku = 18500
jumlah_buku = 7
diskon_persen = 10
uang_budi = 150000

# 1. Total belanja, diskon, dan jumlah yang harus dibayar
total_belanja = harga_buku * jumlah_buku
diskon_rupiah = total_belanja * diskon_persen / 100
jumlah_bayar = total_belanja - diskon_rupiah

# 2. Uang kembalian
uang_kembalian = uang_budi - jumlah_bayar

# 3. Dibagi 4 orang
uang_per_orang = jumlah_bayar // 4
sisa_uang = jumlah_bayar % 4

# 4. Apakah uang Budi cukup?
uang_cukup = uang_budi >= jumlah_bayar

print("========HASIL PERHITUNGAN========")
print("Total       :", int(total_belanja))
print("Diskon      :", int(diskon_rupiah))
print("Bayar       :", int(jumlah_bayar))
print("Kembalian   :", int(uang_kembalian))
print("Per orang   :", int(uang_per_orang))
print("Sisa        :", int(sisa_uang))
print("Uang cukup? :", uang_cukup)
```

Soal 2 : PLTS Atap Rumah
Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

1. Energi yang dihasilkan per hari (kWh).
2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
3. Kekurangan energi per hari terhadap kebutuhan.
4. Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)

```python
jumlah_panel = 12
daya_panel = 450
jam_penyinaran = 5
efisiensi = 0.8
kebutuhan_harian = 25
tarif_listrik = 1444.70

# 1. Energi yang dihasilkan per hari (kWh)
energi_harian = jumlah_panel * daya_panel * jam_penyinaran * efisiensi / 1000

# Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# Kekurangan energi per hari terhadap kebutuhan
kekurangan = kebutuhan_harian - energi_harian

# Apakah PLTS sudah mencukupi kebutuhan rumah?
mencukupi = energi_harian >= kebutuhan_harian

print("========HASIL PERHITUNGAN========")
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

Soal 3 : Generator Diesel dan Rugi Kabel
Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

1. Energi yang dibangkitkan per hari (kWh).
2. Kebutuhan solar per hari dan biayanya.
3. Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
4. Rugi daya pada kabel (P = I² × R).
5. Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)

```python
daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

# 1. Energi yang dibangkitkan per hari (kWh)
energi = daya_generator * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar_harian = energi * konsumsi_solar
biaya_solar = solar_harian * harga_solar

# 3. Berapa hari penuh tangki cukup dan sisa solar
hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

# 4. Rugi daya pada kabel (P = I² × R)
rugi_kabel = arus ** 2 * hambatan

# 5. Apakah rugi kabel tergolong tinggi?
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("========HASIL PERHITUNGAN========")
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar_harian)
print("Biaya (Rp)       :", biaya_solar)
print("Tangki cukup     :", hari_penuh, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
