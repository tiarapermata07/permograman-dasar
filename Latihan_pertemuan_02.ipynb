Latihan 4 — List (sedang)

Berikut daftar pembacaan tegangan (V) sebuah panel:

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

Tugas:

(a) cetak pembacaan pertama dan terakhir memakai indeks

(b) tambahkan satu pembacaan baru 222.0 dengan append()

(c) cetak jumlah total data dengan len().

```python
# Jawaban Latihan 4


tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]


print("tegangan ke-1:", tegangan_harian[0])
print("tegangan ke-6:", tegangan_harian[5])


tegangan_harian.append(222.0)


print("setelah ditambah:", tegangan_harian)


tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

print(len(tegangan_harian))
```

Latihan 5 — Dictionary (sedang)

Cetak salah satu nilai lewat key-nya, lalu ubah nilai rpm (pakai =), dan cetak ulang seluruh dictionary.

Latihan 6 — Gabungan (menantang)

Kelola data tiga unit pembangkit sebagai list of dictionary, tiap unit berisi nama, kapasitas_MW, dan status.

Tugas: cetak nama unit kedua, ubah status salah satu unit, tambahkan satu unit baru ke list dengan append(), lalu cetak seluruh data.

```python
# Jawaban Latihan 6


pembangkit = [{"nama": "Unit-1", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 70, "status": "pemeliharaan"}]


# Cetak nama unit kedua

print(pembangkit[1]["nama"])


# Ubah status salah satu unit

pembangkit[0]["status"] = "pemeliharaan"


# Tambahkan satu unit baru

pembangkit.append({"nama": "Unit-4",
    "kapasitas_MW": 120,
    "status": "operasi"})


# Cetak seluruh data

print(pembangkit)
```
